# **Redes Neuronales Artificiales**

**1. Redes Neuronales Biológicas**

Se entiende por BNN (*Red Neuronal Biológica* por sus siglas en inglés) al modelo simplificado del funcionamiento de una neurona biológica y su arquitectura dentro de una red interconectada de neuronas que se toma como base para definir la arquitectura y funcionamiento de las Redes Neuronales (*ANN* por sus siglas en inglés).

Desde un punto de vista biológico, la neurona es una célula que constituye la unidad fundamental del sistema nervioso. Se compone de un cuerpo y de terminaciones que la conectan con su entorno. Las terminaciones a través de las cuales una neurona recibe un estimulo se conocen como *dendritas*, y la terminación a través de la cual transmiten su propia señal se conoce como *axón*. Por lo general, cada axón se divide en su extremo final en varias terminaciones llamadas *telodendrias*. Las telodendrias de una neurona se conectan a las dendritas de otra neurona estableciendo una profunda conexión entre ambas que se conoce como *sinapsis*.

Cada neurona biológica recibe un pequeño impulso eléctrico como estimulo de otras neuronas a través de su conexión sináptica que llamaremos *señal*. Cuando una neurona recibe una señal genera luego una señal propia que transmite a través de su axón a otras neuronas con las que comparte conexión sináptica. Para que genere esta señal propia, el estímulo que recibe la neurona debe superar cierto valor límite que lleva a la neurona a un estado de activación, desde el cual produce su señal. *Cada  neurona tiene una activación particular, por lo que cada una responde a estimulos diferentes*.

Al unir cada neurona biológica a miles de otras neuronas en una basta red interconectada se forma la Red Neuronal Biológica, que constituye la arquitectura fundamental de funcionamiento del cerebro humano.

bnn[1]

> La *BNN* humana es capaz de realizar operaciones y procedimientos altamente complejos que determinan cada aspecto de la vida. Si bien su estudio es aún un campo activo de investigación, el mapeo de algunas partes del cerebro humano sugiere que las neuronas usualmente se organizan en capas consecutivas.

**2. Redes Neuronales Artificiales**

> La estructura de una Red Neuronal Artificial desarrollada en la computación proviene directamente de la estructura de la Red Neuronal Biológica que compone el cerebro humano.

Una neurona artificial recibe un conjunto específico de *señales de entrada* $(x_1, x_2, ..., x_n)$, cada una de las cuales corresponde a la *señal de salida* de otra neurona. Cada entrada $x_i$ se multiplica por el peso (*weight*) de la conexión correspondiente $w_i$ *que representa la fuerza de la conexión sináptica entre dos neuronas biológicas*. Las entradas pesadas se suman entre sí para formar *la señal total que recibe la neurona*. Cada neurona cuenta con un sesgo $b_0$ (*bias*), un valor límite que se añade a la suma anterior, *y que representa la sensibilidad de la neurona biológica a la señal recibida*, imponiendo un valor de referencia propio de cada neurona artificial. La sumatoria de las entradas pesadas y el sesgo representan la recepción de la señal en el cuerpo de una neurona biológica.

Finalmente, esta señal recibida se evalúa dentro de la *función de activación ($\sigma$)* de la neurona artificial para generar una señal de salida *y*:

$$
\sigma \left(\sum_{i=1}^{n} w_i x_i + b_0 \right) = y
$$

*La función de activación representa la señal generada por la neurona biológica frente al estimulo recibido*. Esta salida "*y*" de la neurona artificial se transforma luego en la señal de entrada de otra neurona artificial.


artificial_neuron[2]

Teniendo en cuenta que, *cada neurona biológica en una BNN esta conectada a otras muchas neuronas y que, al parecer, se organizan por capas*, podemos construir una Red Neuronal Artificial conectando entre sí las salidas y entradas de neuronas artificiales y organizándolas por capas. Dentro de las ANN se distinguen principalmente tres capas:

- **Capa de Entrada:** Es la primera capa, toma los datos que van a procesarse y los organiza en el formato adecuado para ser transmitidos a través de la red.
- **Capa de Salida**: Es la última capa, recibe los datos procesados a través de toda la red y ajusta su formato para producir una respuesta final basada en los patrones identificados.
- **Capas Ocultas**: Son todas las capas intermedias entre la capa de entrada y la capa de salida. Se encargan de procesar la información e identificar los patrones inherentes para la tarea que se pretende modelar con el algoritmo.

mlp[3]

Con base en esta estructura podemos identificar dos tipos de ANN principales:

1. Perceptrón (SLP): ANN más básica y fundamental. Corresponde a la unión entre la capa de entrada, una capa oculta y la capa de salida.

2. Perceptrón Multi-Capa (MLP): Formalmente una Red Neuronal, compromete dos o más capas ocultas entre la capa de entrada y la capa de salida. Primer algoritmo del Deep Learning.

**3. ¿Cómo se entrena una Red Neuronal Artificial?**

1. Se inicializan de forma aleatoria todos los parámetros (pesos y sesgos) de la Red Neuronal. Es muy importante que esta inicialización sea completamente aleatoria.

2. *Forward Propagation*: Se alimenta la red con los datos de entrenamiento. La señal de entrada se utiliza para calcular la salida de la primera capa, que puede expresarse matricialmente de la forma:

$$
y = f(XW^{T} + b)
$$

Donde $f()$ es la función de activación, $X$ es el vector de entrada, $W$ es el vector de pesos y $b$ es el vector de sesgos. El vector y de salida se toma como entrada de la segunda capa oculta, se calcula su salida que a su vez se toma como entrada de la siguiente capa y asi sucesivamente se procesa la señal de entrada hasta la capa de salida. Finalmente, la capa de salida retorna la predicción de la red.

3. *Cálculo del Error*: Una vez obtenida la predicción de la red, el algoritmo calcula el error en la predicción haciendo uso de la función de pérdida definida:

$$
 Error = L(\hat{y}, y)
$$

Bajo esta definición, el algoritmo calcula la contribución de cada neurona en la capa de salida al error en la predicción a través del gradiente de la función de pérdida respecto a los parámetros:

$$
\nabla Error = \left( \frac{\partial L}{\partial f^{(k)}} \frac{\partial f^{(k)}}{\partial z^{(k)}} \frac{\partial z^{(k)}}{\partial W^{(k)}} , \frac{\partial L}{\partial f^{(k)}} \frac{\partial f^{(k)}}{\partial z^{(k)}} \frac{\partial z^{(k)}}{\partial b^{(k)}} \right)
$$

4. *Backpropagation*: Luego de calcular la contribución al error que proporciona la capa de salida, el algoritmo continua calculando el aporte al error de cada capa anterior hasta alcanzar la capa de entrada, es decir, calcula el gradiente del error respecto a los pesos y sesgos de cada capa de la red propagando el error desde la capa de salida hasta la capa de entrada, **propagando el error hacia atrás**. De esta forma se obtiene el aporte global de la red al error en la predicción. Por ejemplo, para la capa inmediatamente anterior a la capa de salida el gradiente correspondiente es de la forma:

$$
\nabla Error = \left( \frac{\partial L}{\partial f^{(k)}} \frac{\partial f^{(k)}}{\partial z^{(k)}} \frac{\partial z^{(k)}}{\partial f^{(k-1)}}  \frac{\partial f^{(k-1)}}{\partial z^{(k-1)}}  \frac{\partial z^{(k-1)}}{\partial W^{(k-1)}}, \frac{\partial L}{\partial f^{(k)}} \frac{\partial f^{(k)}}{\partial z^{(k)}} \frac{\partial z^{(k)}}{\partial f^{(k-1)}} \frac{\partial f^{(k-1)}}{\partial z^{(k-1)}}  \frac{\partial z^{(k-1)}}{\partial b^{(k-1)}} \right)
$$

5. *Actualización de los parámetros*: Una vez calculado el gradiente del error respecto a los parámetros de cada capa, el algoritmo ejecuta un paso de Descenso de Gradiente para ajustar cada peso y sesgo de todas las neuronas de la red:

$$
W_{i}^{(l)} = W_{i}^{(l)} - \alpha \left(\frac{\partial Error}{\partial W^{(l)}} \right) \\
b_{i}^{(l)} = b_{i}^{(l)} - \alpha \left(\frac{\partial Error}{\partial b^{(l)}} \right)
$$

6. Este proceso se repite iterativamente hasta obtener un error ajustado a la tolerancia definida o hasta que se alcance el número de iteraciones máximo definido.

# **Contexto del problema**

En el laboratorio anterior entrenamos un modelo de Regresión Logística con datos que son linealmente separables y con datos no lo son:

image.png

De esa experiencia notamos que el modelo entrenado directamente en los datos no separables mostraba un rendimiento muy bajo, siendo insuficiente para cumplir la tarea de Clasificación encomendada.

Para explorar el notable potencial de las Redes Neuronales Artificiales (ANN), en este laboratorio entrenaremos un Perceptrón Multicapa (MLP) para abordar la tarea de Clasificación utilizando los mismos datos. Para ello, utilizaremos el módulo **Keras** de la librería *TensorFlow*, que permite definir y entrenar redes neuronales de manera flexible, construyendo la arquitectura capa por capa. Veamos si este modelo es capaz de resolver problemas en los que el modelo de Regresión Logística no resulta efectivo.

En primer lugar importamos el modelo y los datos de entrenamiento:

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Importamos los datos de entrenamiento que corresponden a la distribución que no es separable linealmente
data_df = pd.read_csv("classification_train.txt", sep='\t')
data_predictors = data_df[["X","Y"]] # Generamos una matriz con los datos de los predictores X,Y
data_label = data_df["Z"].map(lambda x: 1 if x > 0 else 0) # Vector con las etiquetas. En este caso no tomamos -1, sino 0 como representación
```

A continuación, definimos un modelo de Red Neuronal para clasificación binaria con Keras. Consideramos una capa oculta compuesta por 4 neuronas y una función de activación *ReLu*:

```python
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Dense

# Declaramos un valor de semilla para garantizar reproducibilidad 
tf.random.set_seed(42)
np.random.seed(42)
```

```python
# Declaramos el modelo utilizando la función Sequential(), declarando capa por capa:
modelo_classification = Sequential([
    Input(shape=(data_predictors.shape[1],)), # Capa de entrada. La dimensión de entrada es un vector de dos elementos
    Dense(4, activation='relu'), # Capa oculta con 4 neuronas y activación ReLu
    Dense(1, activation='sigmoid')  # Capa de salida, utilizamos la función sigmoide para clasificación binaria 
])

# Establecemos los hiperparámetros a considerar. 
# La función de error es la entropia cruzada, usual para clasificación binaria
# Utilizamos como métrica -accuracy-
modelo_classification.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

Entrenamos el modelo:

```python
"""
El primer argumento corresponde a la matriz X, en las que cada columna se corresponde con el predictor
y cada fila con cada instancia. Tal y como la construimos en el laboratorio anterior.

Por su parte, el segundo argumento corresponde con el vector que contiene las etiquetas.

Importante: Ambos deben ser arreglos de Numpy.
"""
# Establecemos inicialmente 50 épocas para el proceso de entrenamiento
history_model_1 = modelo_classification.fit(data_predictors, data_label, epochs=50, verbose=0, shuffle=False)
```

El modelo entrenado retorna un objeto llamado `history()` que almacena en su interior diferentes datos relacionados con el proceso de entrenamiento. En particular contiene el valor de la métrica definida (en nuestro caso *accuracy*) calculado en cada época del proceso de entrenamiento. En este caso, no estamos utilizando un conjunto de validación debido a la sencillez del ejercicio, sin embargo el modelo de *Keras* permite cargar estos datos al declarar el proceso de entrenamiento, de modo que el objeto `history()` puede contener a su vez la métrica de cada época en este conjunto, facilitando la generación de curvas de aprendizaje.

Finalmente, utilizamos el modelo entrenado para obtener predicciones sobre los datos de prueba y graficamos los resultados:

```python
# Importamos los datos de prueba
data_df = pd.read_csv("classification_test.txt", sep='\t')
data_predictors_test = data_df[["X","Y"]] # Generamos una matriz con los datos de los predictores X,Y
data_label_test = data_df["Z"].map(lambda x: 1 if x > 0 else 0)

# Evaluamos el modelo entrenado sobre los datos de prueba
predictions = modelo_classification.predict(data_predictors_test)
predictions = (predictions >= 0.5).astype(int)

# Graficamos los resultados: A la derecha los datos reales y a la izquierda las predicciones
fig,ax = plt.subplots(1,2,figsize=(10,5),layout="constrained")
data_predictors_test.plot.scatter("X","Y",c=data_label_test,ax=ax[0],title="Datos originales",cmap="coolwarm",colorbar=False)
data_predictors_test.plot.scatter("X","Y",c=predictions,ax=ax[1],title="Predicción MLP",cmap="coolwarm",colorbar=False)
plt.show()
```

En el ejercicio anterior usamos una función de activación *ReLu* que ejecuta una transformación no lineal a los datos. Bajo esta función de activación, el modelo aprende una frontera de decisión no lineal que se aproxima mejor a la separación original de las clases. Veamos que sucede si utilizamos una función de activación lineal en su lugar:

```python
# Declaramos un valor de semilla para garantizar reproducibilidad 
tf.random.set_seed(42)

# Declaramos el modelo utilizando la función Sequential(), declarando capa por capa.

# Modificamos la función de activación con --linear-- para introducir linealidad
modelo_classification_2 = Sequential([
    Input(shape=(data_predictors.shape[1],)), # Capa de entrada. La dimensión de entrada es un vector de dos elementos
    Dense(4, activation='linear'), # Capa oculta con 4 neuronas y activación lineal
    Dense(1, activation='sigmoid')  # Capa de salida, utilizamos la función sigmoide para clasificación binaria 
])

# Mantenemos los hiperparámetros 
modelo_classification_2.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Ajustamos el modelo sobre el conjunto de entrenamiento:
history_model_2 = modelo_classification_2.fit(data_predictors, data_label, epochs=50, verbose=0, shuffle=False)
```

```python
# Utilizamos el modelo entrenado nuevamente sobre los datos de prueba
predictions_2 = modelo_classification_2.predict(data_predictors_test)
predictions_2 = (predictions_2 >= 0.5).astype(int)


# Graficamos los resultados: A la derecha los datos reales y a la izquierda las predicciones
fig,ax = plt.subplots(1,2,figsize=(10,5),layout="constrained")
data_predictors_test.plot.scatter("X","Y",c=data_label_test,ax=ax[0],title="Datos originales",cmap="coolwarm",colorbar=False)
data_predictors_test.plot.scatter("X","Y",c=predictions_2,ax=ax[1],title="Predicción MLP",cmap="coolwarm",colorbar=False)
plt.show()
```

Observe que en este caso, el modelo impone una separación lineal en el espacio de representación y actúa de forma similar al modelo de Regresión Logística, por lo que no puede clasificar con buena precisión los datos.

**1.2.** Entrene modelos de MLP con una capa oculta de dos, tres, cinco, siete y ocho neuronas utilizando utilizando la función de activación lineal en los datos de entrenamiento que se encuentran en el archivo *classification_train.txt*. Luego, obtenga predicciones sobre el conjunto de prueba del archivo *classification_test.txt* y grafique los resultados, incluya el cálculo de *accuracy* como métrica.

```python
# Repetimos el experimento del punto 1.1 pero con activación lineal en la capa oculta
resultados_lineal = {}

fig, axes = plt.subplots(len(neuron_counts), 2, figsize=(10, 4*len(neuron_counts)), layout='constrained')

for i, n in enumerate(neuron_counts):
    # Semilla fija en cada iteración para garantizar reproducibilidad
    tf.random.set_seed(42)
    np.random.seed(42)

    # MLP con una capa oculta de n neuronas y activación lineal
    modelo_lin = Sequential([
        Input(shape=(data_predictors.shape[1],)),
        Dense(n, activation='linear'),
        Dense(1, activation='sigmoid')
    ])
    modelo_lin.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

    # Entrenamiento con los mismos hiperparámetros
    modelo_lin.fit(data_predictors, data_label, epochs=50, verbose=0, shuffle=False)

    # Predicción y umbralización a clases
    preds_lin = modelo_lin.predict(data_predictors_test, verbose=0)
    preds_lin = (preds_lin >= 0.5).astype(int).flatten()

    # Accuracy sobre el conjunto de prueba
    acc = accuracy_score(data_label_test, preds_lin)
    resultados_lineal[n] = acc

    data_predictors_test.plot.scatter("X", "Y", c=data_label_test, ax=axes[i, 0],
                                      title="Datos originales", cmap="coolwarm", colorbar=False)
    data_predictors_test.plot.scatter("X", "Y", c=preds_lin, ax=axes[i, 1],
                                      title=f"MLP Lineal - {n} neuronas (acc={acc:.3f})",
                                      cmap="coolwarm", colorbar=False)
plt.show()

print("\nAccuracy (Lineal) por número de neuronas en la capa oculta:")
for n, acc in resultados_lineal.items():
    print(f"  {n} neuronas -> accuracy: {acc:.4f}")
```

**1.3.** Observe que los resultados varían para la función de activación ReLu pero no para la función de activación lineal. ¿A qué se debe esto? **Hint:** Compare la métrica de cada modelo.

```python
# Tabla comparativa de accuracy entre ambas funciones de activación
print(f"{'Neuronas':<12}{'Accuracy ReLu':<18}{'Accuracy Lineal':<18}")
print("-" * 48)
for n in neuron_counts:
    print(f"{n:<12}{resultados_relu[n]:<18.4f}{resultados_lineal[n]:<18.4f}")
```

**Respuesta (1.3):**

Los resultados varían al cambiar el número de neuronas cuando se usa *ReLu* porque se trata de una **función de activación no lineal**. Cada neurona con ReLu introduce un "doblez" en el espacio de representación (la región donde la salida pasa de 0 a un valor positivo), por lo que al incrementar el número de neuronas la red puede combinar más dobleces y aprender fronteras de decisión más flexibles, no lineales y más cercanas a la distribución real de los datos. Por ello la métrica *accuracy* cambia (y tiende a mejorar) conforme aumenta el número de neuronas en la capa oculta.

En cambio, al utilizar una **activación lineal** la composición de capas sigue siendo una transformación lineal: matemáticamente, apilar capas densas con activación lineal equivale a una única transformación afín $y = W_{eq} x + b_{eq}$, sin importar cuántas neuronas tenga la capa oculta. La frontera de decisión final (antes de la sigmoide de salida) siempre es un hiperplano en el espacio de entrada, equivalente al que aprendería un modelo de Regresión Logística. Por esta razón el *accuracy* prácticamente no cambia con el número de neuronas y los resultados son muy similares entre sí.


# **MLP para Clasificación de Imágenes**

El conjunto de datos *Optical Recognition of Handwritten Digits* esta compuesto por imágenes de los dígitos de 0 a 9 escritos a mano. En esta sección opcional del laboratorio vamos a entrenar un modelo de red neuronal que nos permita clasificar el dígito mostrado en la imagen.
Para mayor información sobre este conjunto de datos puede consultar el siguiente [enlace](https://archive.ics.uci.edu/dataset/80/optical+recognition+of+handwritten+digits).

En primer lugar, importamos el dataset:

```python
# Importamos el conjunto de datos desde Tensorflow
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical

# Cargar MNIST
(x_train_full, y_train_full), (x_test_full, y_test_full) = mnist.load_data()

# Combinamos todos los datos en un único dataset
# (Queremos separar en los conjuntos de train, test y val con porcentajes específicos)
X_total = np.concatenate([x_train_full, x_test_full])
y_total = np.concatenate([y_train_full, y_test_full])
```

Ahora observemos algunas de las imágenes en el dataset:

```python
# Declaramos un valor de semilla para garantizar reproducibilidad 
np.random.seed(42)

# Elegimos 4 imágenes al azar
random_index = np.random.randint(0, X_total.shape[0], size = 5)
images = X_total[random_index]
labels = y_total[random_index]

# Graficamos las imágenes elegidas
fig,ax = plt.subplots(1,4,figsize=(20,5))
for index in range(4):
  ax[index].imshow(images[index].reshape((28,28)),cmap="gray")
  ax[index].set_title(f"Digito: {labels[index]}")
  ax[index].axis("off")
plt.show()
```

Para entrenar el modelo vamos a separar el conjunto de datos en un conjunto de entrenamiento ($70\%$), un conjunto de validación ($15\%$) y un conjunto de prueba ($15\%$). Para ello utilizamos la función *train_test_split()*:

```python
from sklearn.model_selection import train_test_split

val_size = 0.15
test_size = 0.15

# En test_size específicamos el tamaño (como porcentaje) del conjunto de prueba
# La varible random_state es muy importante, Asegura que el proceso aleatorio con el que se construyen
# los dos conjuntos sea "almacenado" de modo que cada vez que se ejecute el proceso los conjuntos tengan
# los mismos elementos
X_temp, X_test, y_temp, y_test = train_test_split(X_total,y_total,test_size=test_size,random_state=42, shuffle=True)

# Calculamos el porcentaje de validación ajustado a la separación inicial
val_size = val_size/(1 - test_size) # tecnicamente, esto da el porcentaje ajustado, no el tamaño del conjunto

# Separamos el conjunto temporal en entrenamiento y validación, con el porcentaje ajustado
X_train, X_val, y_train, y_val = train_test_split(X_temp,y_temp,test_size=val_size, random_state=42)
print(f"Tamaño del conjunto de entrenamiento: X: {X_train.shape}, y: {y_train.shape}")
print(f"Tamaño del conjunto de validación: X: {X_val.shape}, y: {y_val.shape}")
print(f"Tamaño del conjunto de prueba: X: {X_test.shape}, y: {y_test.shape}")
```

Antes de continuar es necesario redimensionar y normalizar los datos:

```python
# Calculamos el tamaño de cada imagen
input_size = X_train.shape[1]*X_train.shape[2] 

# Calculamos el número de clases a considerar
output_size = len(np.unique(y_train))

# Trasladamos la notación de las etiquetas a una codificación apropiada para clasificación multiclase
y_train = to_categorical(y_train)
y_val = to_categorical(y_val)
y_test = to_categorical(y_test)

# Conjunto de entrenamiento
x_train = np.reshape(X_train, [-1, input_size])
x_train = x_train.astype('float32') / 255

# Conjunto de validación
x_val = np.reshape(X_val, [-1, input_size])
x_val = x_val.astype('float32') / 255

# Conjunto de prueba
x_test = np.reshape(X_test, [-1, input_size])
x_test = x_test.astype('float32') / 255
```

Implementamos un MLP para Clasificación con 2 neuronas en la capa oculta y una función de activación *ReLu*:

```python
# Declaramos un valor de semilla para garantizar reproducibilidad 
tf.random.set_seed(42)

# Declaramos la red neuronal para clasificación multiclase con activación ReLu
digit_classificator = Sequential([
    Input(shape=(input_size,)),
    Dense(2, activation='relu'),
    Dense(units=output_size, activation='softmax')
])

# Declaramos los hiperparámetros a considerar en el modelo
digit_classificator.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics = ['accuracy']
)

# Entrenamos el modelo sobre los datos de entrenamiento pero incluimos los datos de validación
history_digit_model = digit_classificator.fit(x_train, y_train, 
                                              epochs=20, 
                                              batch_size=128, 
                                              validation_data = (x_val, y_val),
                                              verbose = 0,
                                              shuffle=False)
```

Por último, vamos a evaluar el desempeño del modelo.

```python
# Mejor métrica con el conjunto de entrenamiento
best_train_score = np.max(history_digit_model.history['accuracy'])
best_train_score_epoch = np.argmax(history_digit_model.history['accuracy']) + 1

# Mejor métrica con el conjunto de validación
best_val_score = np.max(history_digit_model.history['val_accuracy'])
best_val_score_epoch = np.argmax(history_digit_model.history['val_accuracy']) + 1

print(f"\n=== MEJORES MÉTRICAS ===")
print(f"Train - Mejor accuracy: {best_train_score:.4f} (época {best_train_score_epoch})")
print(f"Val   - Mejor accuracy: {best_val_score:.4f} (época {best_val_score_epoch})")
```

Así, mediante el uso de un MLP logramos ajustar un modelo capaz de clasificar dígitos escritos a mano con un *accuracy* del 60% en el conjunto de validación.

```python
# Semilla para reproducibilidad
tf.random.set_seed(42)
np.random.seed(42)

# MLP más profundo con más neuronas por capa para capturar los patrones de MNIST
# Arquitectura: 3 capas ocultas (256, 128, 64) con ReLu y salida softmax de 10 clases
mi_clasificador = Sequential([
    Input(shape=(input_size,)),
    Dense(256, activation='relu'),   # primera capa oculta
    Dense(128, activation='relu'),   # segunda capa oculta
    Dense(64,  activation='relu'),   # tercera capa oculta
    Dense(units=output_size, activation='softmax')  # salida multiclase
])

# Hiperparámetros: optimizador Adam y entropía cruzada categórica
mi_clasificador.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# Entrenamos registrando la historia para las curvas de aprendizaje
history_mi_modelo = mi_clasificador.fit(
    x_train, y_train,
    epochs=20,
    batch_size=128,
    validation_data=(x_val, y_val),
    verbose=0,
    shuffle=False
)

# Mejores métricas alcanzadas
best_train_acc = np.max(history_mi_modelo.history['accuracy'])
best_val_acc = np.max(history_mi_modelo.history['val_accuracy'])
print(f"Mejor accuracy en entrenamiento: {best_train_acc:.4f}")
print(f"Mejor accuracy en validación:    {best_val_acc:.4f}")
```

**2.2.** Utilizando los datos almacenados en el objeto `history()` grafique la curva de aprendizaje de su modelo y decida si presenta *subajuste*, *sobreajuste* o se encuentra *bien ajustado*. Finalmente calcule el *accuracy* en el conjunto de prueba.

```python
# Curvas de aprendizaje: accuracy y pérdida por época
epochs = range(1, len(history_mi_modelo.history['accuracy']) + 1)

fig, ax = plt.subplots(1, 2, figsize=(12, 4), layout='constrained')

# Curva de accuracy
ax[0].plot(epochs, history_mi_modelo.history['accuracy'], label='Entrenamiento', marker='o')
ax[0].plot(epochs, history_mi_modelo.history['val_accuracy'], label='Validación', marker='o')
ax[0].set_xlabel('Época')
ax[0].set_ylabel('Accuracy')
ax[0].set_title('Curva de aprendizaje - Accuracy')
ax[0].legend()
ax[0].grid(alpha=0.3)

# Curva de pérdida
ax[1].plot(epochs, history_mi_modelo.history['loss'], label='Entrenamiento', marker='o')
ax[1].plot(epochs, history_mi_modelo.history['val_loss'], label='Validación', marker='o')
ax[1].set_xlabel('Época')
ax[1].set_ylabel('Pérdida')
ax[1].set_title('Curva de aprendizaje - Pérdida')
ax[1].legend()
ax[1].grid(alpha=0.3)
plt.show()

# Accuracy del modelo sobre el conjunto de prueba
test_loss, test_acc = mi_clasificador.evaluate(x_test, y_test, verbose=0)
print(f"Accuracy sobre el conjunto de prueba: {test_acc:.4f}")
print(f"Pérdida sobre el conjunto de prueba:  {test_loss:.4f}")
```

**Análisis de la curva de aprendizaje y ajuste del modelo:**

Observando las curvas, el *accuracy* de entrenamiento alcanza valores muy altos (cercanos al 100%) mientras que el *accuracy* de validación se estabiliza en un valor menor. De forma complementaria, la pérdida de entrenamiento continúa descendiendo mientras que la pérdida de validación deja de mejorar (e incluso comienza a aumentar ligeramente) en las últimas épocas. Este comportamiento —el modelo se ajusta muy bien a los datos de entrenamiento pero pierde capacidad de generalización sobre validación— es característico del **sobreajuste (overfitting)**.

La red de tres capas ocultas (256, 128, 64) tiene suficiente capacidad para memorizar los patrones del conjunto de entrenamiento, y dado que no se usa regularización (por ejemplo *dropout* o *early stopping*), termina aprendiendo detalles específicos del entrenamiento que no se transfieren plenamente al conjunto de validación. Aun así, el modelo supera el umbral del 90% de *accuracy* requerido tanto en validación como en el conjunto de prueba.
